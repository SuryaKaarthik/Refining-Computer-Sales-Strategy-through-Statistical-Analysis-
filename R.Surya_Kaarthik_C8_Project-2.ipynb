### Task 1

The population standard deviation is known and sample > 30 
So we need to perform z test
It is one sample test
The type is one tail right 

```python
population_mean = 221.3
population_stdev = 74.83
```

```python
sample_mean = 222.2
n = 40
```

```python
alpha=0.05
```

```python
z_statistics = (sample_mean - population_mean) / (population_stdev / math.sqrt(n))
```

```python
p_value = norm.sf(abs(z_statistics))
```

```python
z_critical = norm.ppf(1 - alpha)
```

```python
print("The z_statistics is ", z_statistics)
print("The p_value is ", p_value)
print("The z_critical is ", z_critical)
```

### Task-2

```python
computers_with_CD_players = df.loc[df["cd"] == "yes"]
computers_with_CD_players
```

```python
computers_with_no_CD_players = df.loc[df["cd"] == "no"]
computers_with_no_CD_players
```

```python
mean1 = st.mean(computers_with_CD_players["price"])
stdev1 = st.stdev(computers_with_CD_players["price"])
```

```python
mean2 = st.mean(computers_with_no_CD_players["price"])
stdev2 = st.stdev(computers_with_no_CD_players["price"])
```

```python
shape1 = computers_with_CD_players.shape
shape2 = computers_with_no_CD_players.shape
```

```python
n1 = shape1[0]
n2 = shape2[0]
```

```python
print("The mean of the computer with cd player is ", mean1)
print("The stdev of the computer with cd player is ", stdev1)
print("The mean of the computer with no cd player is ", mean2)
print("The stdev of the computer with no cd player is ", stdev2)
print("The total sample of computer with cd player is ", n1)
print("The total sample of computer with no cd player is ", n2)
```

```python
alpha = 0.05
```

```python
z_statistics = (mean1 - mean2) / math.sqrt(((stdev1 ** 2) / n1) + ((stdev2 ** 2) / n2))
```

```python
p_value = norm.sf(abs(z_statistics))*2
```

```python
z_critical = norm.ppf(1 - alpha / 2)
```

```python
print("The z_statistics is ", z_statistics)
print("The p_value is ", p_value)
print("The z_critical is ", z_critical)
```

### Task-3

```python
H0 : the mean price of premium computers is $2200
H1 : the mean price of premium computers differs significantly from $2200
```

```python
population_mean = 2200
```

```python
alpha = 0.05
```

```python
premium_sample = df.loc[df["premium"] == "yes"]
```

```python
premium_sample_data = premium_sample.sample(n = 25, random_state = 1, replace = True)
```

```python
n = 25
dof = n - 1
```

```python
sample_mean = st.mean(premium_sample_data["price"])
sample_stdev = st.stdev(premium_sample_data["price"])
```

```python
t_statistics = (sample_mean - population_mean ) / (sample_stdev / math.sqrt(n))
```

```python
p_value = t.sf(abs(t_statistics), dof)*2
```

```python
t_critical = t.ppf(1-alpha / 2, dof)
```

```python
print(f"The t_statistics is ", t_statistics)
print(f"The p_value is ", p_value)
print(f"The t_critical is ", t_critical)
```
