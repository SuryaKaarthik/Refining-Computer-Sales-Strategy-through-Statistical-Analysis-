PDF (Probability Density Function)
In statistics, PDF stands for Probability Density Function, which is used to describe the distribution of continuous random variables.

It applies specifically to continuous data.

The x-axis represents the data values, and the y-axis shows the likelihood of those values.

Rather than giving the exact probability of a specific value, the PDF indicates how likely a value is relative to others.

Peaks (higher points) on the PDF curve represent values that are more probable.

Lower parts of the curve correspond to less likely values.

The probability that a value lies within a certain interval is represented by the area under the curve between those two points.

The total area under the entire PDF curve is always equal to 1, which accounts for all possible outcomes.
